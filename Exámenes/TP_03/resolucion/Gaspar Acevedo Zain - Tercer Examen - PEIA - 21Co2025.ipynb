### Resolución Ejercicio número 1

Para determinar los intervalos de confianza empíricos para el supermercado `Santa Ana` en cada mes con significancias del 95% y 99%, debemos primero determinar los siguientes conceptos:

1. ¿Es un intervalo bilateral, unilateral superior, o unilateral inferior?
    1. Debido a que buscamos definir intervalos de confianza para las ventas de un supermercado dado, lo que nos interesa saber son los valores mínimos y máximos que pueden llegar a tener, motivo por el cual utilizaremos un `intervalo bilateral`.
1. ¿Qué distribución utilizar?
    1. Por un lado, no tenemos información de la media y varianza de la muestra, solamente tenemos observaciones, más precisamente, entre 30 y 31 observaciones por mes (a excepción de Febrero, que son 28), una por cada día.
    1. Por este motivo, como no tenemos información de la varianza, pero si tenemos más de 30 observaciones (salvo para el mes de Febrero), decidimos utilizar una distribución `Normal`, por lo cual, utilizaremos la media muestral y la varianza muestral para realizar el cálculo de nuestros intervalos de confianza.
    1. El motivo es que, si bien cuando no se conoce la varianza suele utilizarse una distribución `t-student`, como los tamaños de muestras son mayores a 30 para cada mes, la `t-student` suele comportarse como una `Normal`
1. A partir de los valores de significancia, ¿Cuál es el punto crítico de cada uno?
    1. Primero, definimos nuestros alfas de la siguiente manera: $p = 1 - \alpha$, siendo $p_{1} = 0.95$ y $p_{2} = 0.99$ para los respectivos niveles de significancia, quedando $\alpha_1 = 0.05$ y $\alpha_2 = 0.01$
    1. Ahora bien, a la hora de determinar los valores críticos para cada intervalo mediante una distribución `Normal`, como estamos usando un intervalo a dos colas, debemos dividir los alfa por dos, quedando: $\alpha_1 = 0.025$ y $\alpha_2 = 0.005$

Por lo dicho anteriormente, nuestros intervalos de confianza quedan definidos de la siguiente manera:

$P(\bar{x}-\frac{k^*\sigma}{\sqrt{n}}; \bar{x}+\frac{k^*\sigma}{\sqrt{n}})$

Siendo:
- $\bar{x}$ la media muestral, la cual calcularemos a partir de los datos dados
- $n$ el tamaño de la muestra
- $\sigma$ el desvío estandar, el cual calcularemos a partir de la varianza muestral $\sigma^{2}$
- $k^{*}$ o el valor crítico, el cual obtendremos mediante las tablas de la distribución `Normal` y los alfas definidos

### Código para la resolución del Ejercicio número 1

```python
# Esta función nos permite obtener un intervalod de confianza a partir de tres datos:
# datos: las observaciones en sí
# nivel de confianza: el nivel de confianza, expresado en decimales
# bilateral: es el intervalo bilateral, o no?
def intervalo_confianza(datos, nivel_de_confianza, bilateral = True):
    alfa = 1 - nivel_de_confianza
    if bilateral:
        alfa = alfa/2

    varianza_muestral = statistics.variance(datos) # utiliza el denominador (N-1)
    desvio_muestral = math.sqrt(varianza_muestral)
    media_muestral = statistics.mean(datos)
    raiz_n = math.sqrt(len(datos))
    k = st.norm.ppf(1-alfa)

    intervalo_valor_izq = media_muestral - k*desvio_muestral/raiz_n
    intervalo_valor_der = media_muestral + k*desvio_muestral/raiz_n

    return intervalo_valor_izq, intervalo_valor_der, k
```

```python
columnas_95=["Intervalo (95%)", "Limite_Inferior", "Limite_Superior", "Valor_Critico Utilizado"]
columnas_99=["Intervalo (99%)", "Limite_Inferior", "Limite_Superior", "Valor_Critico Utilizado"]
df_95 = pd.DataFrame(index=meses, columns=columnas_95)
df_99 = pd.DataFrame(index=meses, columns=columnas_99)
```

```python
for i in range(1,13):
    df_por_mes = df_santa_ana[df_santa_ana["FechaMes"] == i]
    data = df_por_mes.Ventas

    nivel_confianza_01 = 0.95
    nivel_confianza_02 = 0.99

    limite_inferior_095, limite_superior_095, k_095 = intervalo_confianza(data, nivel_confianza_01, True)
    limite_inferior_099, limite_superior_099, k_099 = intervalo_confianza(data, nivel_confianza_02, True)

    df_95.loc[nombre_meses[i]] = [f"({round(limite_inferior_095, 2)}, {round(limite_superior_095, 2)})", round(limite_inferior_095, 2), round(limite_superior_095, 2), round(k_095, 2)]
    df_99.loc[nombre_meses[i]] = [f"({round(limite_inferior_099, 2)}, {round(limite_superior_099, 2)})", round(limite_inferior_099, 2), round(limite_superior_099, 2), round(k_099, 3)]
```
