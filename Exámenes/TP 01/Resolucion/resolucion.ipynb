```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_ventas = pd.read_excel('Datos_primer_TP_21Co2025_a2101.xlsx')
ventas_totales = df_ventas.Ventas

# Dividimos por año
df_ventas["FechaAño"] = df_ventas["Fecha"].dt.year

# Creamos un DataFrame por cada año
df_2021 = df_ventas[df_ventas["FechaAño"] == 2021]
ventas_2021 = df_2021.Ventas
df_2022 = df_ventas[df_ventas["FechaAño"] == 2022]
ventas_2022 = df_2022.Ventas
df_2023 = df_ventas[df_ventas["FechaAño"] == 2023]
ventas_2023 = df_2023.Ventas

# Validamos que la suma de los DF de cada año sea igual al total
len(ventas_totales) == (len(ventas_2021) + len(ventas_2022) + len(ventas_2023))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

axes[0].set_title('Función Empírica de Distribución de Ventas Totales')
axes[0].set_xlabel("Ventas")
axes[0].set_ylabel("Proporción")
sns.ecdfplot(ventas_totales, ax=axes[0])

axes[1].set_title('Función de Densidad de Ventas Totales')
axes[1].set_xlabel("Ventas")
axes[1].set_ylabel("Proporción")
sns.kdeplot(ventas_totales, ax=axes[1])
```

```python
# Empirical Commutative Distribution Function
# https://seaborn.pydata.org/generated/seaborn.ecdfplot.html
# https://www.geeksforgeeks.org/how-to-make-ecdf-plot-with-seaborn-in-python/
sns.ecdfplot(ventas_totales)
# Obtenemos los ejes para definir el Título, y los nombres de los ejes X e Y
axis_totales = plt.gca()
axis_totales.set_title('Función Empírica de Distribución de Ventas Totales')
axis_totales.set_xlabel("Ventas")
axis_totales.set_ylabel("Proporción")
# Graficamos
plt.show()
```

```python
# https://seaborn.pydata.org/generated/seaborn.kdeplot.html
# https://www.geeksforgeeks.org/seaborn-kdeplot-a-comprehensive-guide/
# Función Densidad Probabilidad
sns.kdeplot(ventas_totales)
# Obtenemos los ejes para definir el Título, y los nombres de los ejes X e Y
axis_totales = plt.gca()
axis_totales.set_title('Función de Densidad de Ventas Totales')
axis_totales.set_xlabel("Ventas")
axis_totales.set_ylabel("Proporción")
# Graficamos
plt.show()
```
