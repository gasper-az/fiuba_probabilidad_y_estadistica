## Ejercicio número 1

### Enunciado

(3 puntos) De 10 monedas hay 7 monedas falsas, que tienen probabilidad 0,2 de mostrar cara al ser lanzadas.

Si se toma una moneda al azar, se lanza 14 veces, y en todas ellas se obtiene cara, ¿qué es más probable, que la moneda elegida sea
justa o que esté cargada?

Dar las probabilidades tanto de que la moneda elegida sea falsa, como de que sea justa.

### Paso 1 - Definir Eventos y sus probabilidades

Definamos los siguientes eventos:

- $C_1$: Sale Cara. Es el complemento a Sale Cruz ($C_2$).
- $C_2$: Sale Cruz. Es el complemento a Sale Cara ($C_1$).
- $M_f$: Moneda Falsa. Complemento de Moneda Justa ($M_j$).
- $M_j$: Moneda Justa. Complemento de Moneda Falsa ($M_f$).

Además, tenemos estos eventos condicionales, según el enunciado:

- $C_1/M_f$: corresponde al evento en el que sale una *Cara* ($C_1$), sabiendo que la moneda es *Falsa* ($M_f$)
- $C_1/M_j$: corresponde al evento en el que sale una *Cara* ($C_1$), sabiendo que la moneda es *Justa* ($M_j$)
- $C_2/M_f$: corresponde al evento en el que sale una *Cruz* ($C_2$), sabiendo que la moneda es *Falsa* ($M_f$)
- $C_2/M_j$: corresponde al evento en el que sale una *Cruz* ($C_2$), sabiendo que la moneda es *Justa* ($M_j$)

A partir del enunciado definimos las siguientes probabilidades:

- $P(M_f) = 0,7$
  - ya que en el enunciado se menciona que hay 7 monedas falsas en un total de 10 monedas, es decir $7/10 = 0,7$
- $P(M_j) = 0,3$
  - al ser $M_j$ el complemento de $M_f$, podemos calcular $P(M_j) = 1 - P(M_f) = 1 - 0,7 = 0,3$
- $P(C_1/M_f) = 0,2$
  - ya que el enunciado menciona `De 10 monedas hay 7 monedas falsas, que tienen probabilidad 0,2 de mostrar cara al ser lanzadas.`
- $P(C_2/M_f) = 0,2$
  - ya que $C_2/M_f$ es el complemento de $C_1/M_f$, por lo que $P(C_2/M_f) = 1 - P(C_1/M_f) = 1 - 0,2 = 0,8$

Por último, definimos las siguientes probabilidades condicionales
- $P(C_1/M_j) = P(C_2/M_j) = 0,5$
  - Asumimos que $C_1/M_j$ y $C_2/M_j$ son equiprobables, ya que la moneda es justa
    - De aquí se tiene que $P(C_1/M_j) = P(C_2/M_j)$
  - Dado a que $C_1/M_j$ y $C_2/M_j$ son complementarios, $P(C_1/M_j) + P(C_2/M_j) = 1$
  - Como $P(C_1/M_j) = P(C_2/M_j)$, tenemos
    - $2*P(C_1/M_j) = 1$ 
    - $P(C_1/M_j)= 1/2 = 0,5$

### Paso 2 - Planteamos el problema

El enunciado indica que
1. Se toma una moneda al azar.
1. Esa misma moneda, se lanza 14 veces, y en cada tirada se obtiene cara.

De aquí podemos asumir y deducir lo siguiente:
1. Las 14 tiradas (o experimentos) son independientes, y en cada una de ellas siempre se usa la misma moneda.
1. Sabemos que el resultado de cada experimento siempre fue Cara
1. Dado lo anterior, nos interesa saber si, sabiendo que el resultado fue Cara, la moneda está cargada o es justa
1. Es decir, se nos pregunta la probabilidad de los eventos
   1. $M_j/C_1$ o *Moneda Justa*, sabiendo que salió *Cara*
   1. $M_f/C_1$ o *Moneda Falsa*, sabiendo que salió *Cara*
1. Como $M_f/C_1$ y $M_j/C_1$, sus probabilidades suman 1
   1. $P(M_f/C_1) + P(M_j/C_1) = 1$

> Para calcular $P(M_f/C_1)$ nos valemos del teorema de Bayes:

$P(M_f/C_1) = \frac{P(C_1 \cap M_f)}{P(C_1)}$

> Análogamente:

$P(M_j/C_1) = \frac{P(C_1 \cap M_j)}{P(C_1)}$

### Paso 3 - Resolvemos la probabilidad condicional $P(M_f/C_1)$ y $P(M_j/C_1)$

> Gracias al Teorema de Bayes sabemos que:

$P(M_f/C_1) = \frac{P(C_1 \cap M_f)}{P(C_1)}$

> $P(C_1)$ se puede expresar de la siguiente manera:

$P(C_1) = P(C_1 \cap M_f) + P(C_1 \cap M_j)$

> Nuevamente, gracias al Teorema de Bayes podemos expresar lo siguiente:

$P(C_1 \cap M_f) = P(M_f) * P(\frac{C_1}{M_f})$

$P(C_1 \cap M_j) = P(M_j) * P(\frac{C_1}{M_j})$

> Es decir, podemos expresar $P(C_1 \cap M_f)$ y $P(C_1)$ mediante probabilidades que obtuvimos en el Paso 2.

$P(C_1) = P(M_f) * P(\frac{C_1}{M_f}) + P(M_j) * P(\frac{C_1}{M_j})$

$P(C_1) = 0,7 * 0,2 + 0,3 * 0,5$

$P(C_1) = 0,14 + 0, 15

$P(C_1) = 0,29$

> Obtuvimos que $P(C_1) = 0,29$

> Además, obtuvimos que $P(C_1 \cap M_f) = 0,14$ y $P(C_1 \cap M_j) = 0,15$

> Calculamos las probabilidades condicionales $P(M_f/C_1)$ y $P(M_j/C_1)$

$P(M_f/C_1) = \frac{P(C_1 \cap M_f)}{P(C_1)} = \frac{0,14}{0,29} = 0,48275862 \approx 0,4827$

$P(M_j/C_1) = \frac{P(C_1 \cap M_j)}{P(C_1)} = \frac{0,15}{0,29} = 0,517241379 \approx 0,5173$

> Obtuvimos que $P(M_f/C_1) \approx 0,4827$ y que $P(M_j/C_1) \approx 0,5173$

### Paso 4 - Calculamos las probabilidades para 14 eventos

> En el paso anterior obtuvimos que $P(M_f/C_1) \approx 0,4827$ y que $P(M_j/C_1) \approx 0,5173$

> Ahora, queremos saber cuales son las probabilidades de repetir cada evento 14 veces.
> Para ello, nos valemos de la fórmula de *Variable Aleatoria Binomial*, ya que lo que nosotros buscamos es la probabilidad de que,
> al repetir 14 eventos independientes, obtengamos 14 éxitos.

$P(k-éxitos) = C_{n,k} * p^{k} * (1-p)^{n-k}$

> Definimos lo siguiente:
- $n$: Cantidad de experimentos independientes, siendo en nuestro caso $14$
- $k$: Cantidad de éxitos, siendo en nuestro caso $14$
- $C_{n,k}$: Combinatorias de $n$, tomadas de a $k$

> A partir de lo anterior, definimos la combinatoria $C_{n,k}$ de esta manera:

$C_{n,k} = \frac{n!}{(n-k)! * k!)} = \frac{14!}{(14-14)! * 14!} = \frac{14!}{1 * 14!} = 1$

> Reemplazando en la fórmula de *Variable Aleatoria Binomial*:

$P(k-éxitos=14) = C_{n,k} * p^{k} * (1-p)^{n-k} = 1 * p^{14} * (1-p)^{14-14} = p^{14} * (1-p)^{0} = p^{14} * 1 = p^{14}$

> Es decir, para el caso de repetir 14 experimentos y obtener 14 éxitos, la probabilidad está dada por $P(k-éxitos=14) = p^{14}$

> Considerando que nuestro caso de éxito sea que la moneda sea falsa, sabiendo que salió Cara, es decir $M_f/C_1$

$p = P(M_f/C_1) \approx 0,4827$
$P(k-éxitos=14) = p^{14} \approx 0,4827^{14} \approx 0,00003734$

> Considerando que nuestro caso de éxito sea que la moneda sea Justa, sabiendo que salió Cara, es decir $M_j/C_1$

$p = P(M_j/C_1) \approx 0,5173$
$P(k-éxitos=14) = p^{14} \approx 0,5173^{14} \approx 0,000098108$

### Paso 5 - Respuesta

Sabiendo que:
- De 10 monedas hay 7 monedas falsas
- La monedas falsas tienen probabilidad 0,2 de mostrar cara al ser lanzadas
- Se tomó una moneda al Azar y se lanzó 14 veces, y en todas ellas se obtuvo cara

> Respuesta: Podemos decir que es más probable que la moneda sea ***Justa***

- La probabilidad que la moneda sea Justa, sabiendo que salió Cara es $\approx 0,5173$
  - $P(M_j/C_1) \approx 0,5173$
  - Si repetimos el evento anterior (que la moneda sea Justa, sabiendo que salió Cara) 14 veces, obtenemos una probabilidad de $\approx 0,000098108$
    - $P(k-éxitos=14) = p^{14} \approx 0,5173^{14} \approx 0,000098108$
- La probabilidad que la moneda sea Falsa, sabiendo que salió Cara es $\approx 0,4827$
  - $P(M_f/C_1) \approx 0,4827$
  - Si repetimos el evento anterior (que la moneda sea Falsa, sabiendo que salió Cara) 14 veces, obtenemos una probabilidad de $\approx 0,00003734$
    - $P(k-éxitos=14) = p^{14} \approx 0,4827^{14} \approx 0,00003734$

## Ejercicio número 2

### Enunciado

(3 puntos) Sean X e Y dos v.a. continuas con densidad conjunta:

$
f_{X,Y} = \begin{cases}
Ky & \text{si } 9x^{2} \le y \le 5x \\
0 & \text{en otro caso}
\end{cases}
$

Encontrar:

a. Determine el Valor de K </br>
b. Encuentre la densidad marginal $f_Y(y)$ de $Y$ </br>
c. Encuentre la densidad condicional $f_{X|Y}(x|y)$ de $X$ dado $Y$ </br>

### Paso 1 - Definimos el rango de X en la función de densidad conjunta

Según el enunciado, tenemos una función de densidad conjunta definida de la siguiente manera:

$
f_{X,Y} = \begin{cases}
Ky & \text{si } 9x^{2} \le y \le 5x \\
0 & \text{en otro caso}
\end{cases}
$

De la misma podemos saber que la variable $Y$ está definida por el rango $9x^{2} \le y \le 5x$, pero no tenemos un rango definido para la variable $X$.

Para hallarlo, primero graficamos ambas funciones, enfocándonos en el área definida por ambas.

> Se realiza gráfico de ambas funciones mediante matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 1)
y1 = 9*x**2
y2 = 5*x

# Create the plot
fig, ax = plt.subplots()

ax.plot(x, y1, label='$9x^2$', color="green")
ax.plot(x, y2, label='$5x$', color="blue")

# Fill the area between the functions
ax.fill_between(x, y1, y2, where=y1 <= y2, color='gray', alpha=0.3)

ax.set_xlabel('x')
ax.set_xlabel('y')
ax.set_title('Intersección de $9x^2$ y $5x$')
ax.legend()
plt.show()
```

Del gráfico anterior podemos deducir que nuestro área de interés es aquel definido por el área **coloreado en gris**, delimitado por $9x^{2}$ y $5x$.

El intervalo de la variable $Y$ es aquel que define el enunciado:

- $9x^{2} \le y \le 5x$

Mientras que al intervalo de $X$ lo podemos calcular al hacer la intersección de las funciones $9x^{2}$ y $5x$:

$9x^{2} = 5x$

$9x^{2} - 5x = 0$

$x(9x - 5) = 0$

De lo que obtenemos lo siguiente:
- $x_1 = 0$
- $x_2 = \frac{5}{9}$

Entonces, la función de Densidad conjunta está definida de esta manera:

$
f_{X,Y} = \begin{cases}
Ky & \text{si } 9x^{2} \le y \le 5x &;& 0 \le x \le \frac{5}{9} \\
0 & \text{en otro caso}
\end{cases}
$

### Paso 2 - Determinamos el valor de K

Por definición, sabemos que la doble integral de la función de Densidad Conjunta entre los intervalos $X$ e $Y$ es igual a $1$, es decir:

$1 = \int_{0}^{\frac{5}{9}}\int_{9x^2}^{5x}Ky\,dy\,dx$

> Comenzamos integrando respecto a Y

$1 = K*\int_{0}^{\frac{5}{9}}\int_{9x^2}^{5x}y\,dy\,dx$

$1 = K*\int_{0}^{\frac{5}{9}}(\frac{y^{2}}{2})|_{9x^2}^{5x}\,dx$

$1 = K*\int_{0}^{\frac{5}{9}}(\frac{(5x)^{2}}{2} - \frac{(9x^2)^2}{2})\,dx$

$1 = K*\int_{0}^{\frac{5}{9}}(\frac{25x^2}{2} - \frac{81x^4}{2})\,dx$

$1 = K * (\frac{25x^3}{6} - \frac{81x^5}{10})|_{0}^{\frac{5}{9}}$

$1 = \frac{625}{2187} * K$

$ K = \frac{2187}{625}$

> Por lo tanto, la Función de Densidad Conjunta de las variables aleatorias continuas X e Y queda definida como:

$
f_{X,Y} = \begin{cases}
\frac{2187}{625}y & \text{si } 9x^{2} \le y \le 5x &;& 0 \le x \le \frac{5}{9} \\
0 & \text{en otro caso}
\end{cases}
$

### Paso 3 - Definimos la Densidad Marginal $f_Y(y)$ de $Y$

La Densidad Marginal $f_Y(y)$ de $Y$ se define como la integral en función de la variable $X$ de la Función de Densidad Conjunta

$
f_{X,Y} = \begin{cases}
\frac{2187}{625}y & \text{si } 9x^{2} \le y \le 5x &;& 0 \le x \le \frac{5}{9} \\
0 & \text{en otro caso}
\end{cases}
$

Eso significa que tomamos a $Y$ como $Variable Libre$, y a $X$ como $Variable Condicionada$.

Para ello, debemos expresar el rango de $Y$ entre dos números, y al de $X$ en funcion de $Y$.

Comenzamos definiendo al rango de $X$ en función de $Y$
Expresamos a las funciones $y=9x^2$ e $y=5x$ en función de $x$, quedando

- Límite inferior del rango de $X$
  - $y=9x^2$
  - $(\frac{y}{9})^{1/2} = x$
  - $x = \frac{y^{1/2}}{3}$
- Límite superior del rango de $X$
  - $y=5x$
  - $x = \frac{y}{5}$

El rango de $Y$ se calcula como la intersección de $x = \frac{y}{5}$ y $x = \frac{y^{1/2}}{3}$

$\frac{y}{5} = \frac{y^{1/2}}{3}$

Elevamos al cuadrado ambos términos para facilitar su resolución

$\frac{y^2}{25} = \frac{y}{9}$

$\frac{y^2}{25} - \frac{y}{9} = 0$

$y*(\frac{y}{25} - \frac{1}{9}) = 0$

De lo que obtenemos lo siguiente:
- $y_1 = 0$
- $y_2 = \frac{25}{9}$

Procedemos a calcular la Densidad marginal de $f_Y(y)$:

$f_Y(y) = \int_{\frac{y}{5}}^{\frac{y^{1/2}}{3}}\frac{2187}{625}y\,dx$

$f_Y(y) = \frac{2187}{625} * \int_{\frac{y}{5}}^{\frac{y^{1/2}}{3}}y\,dx$

$f_Y(y) = \frac{2187}{625} * y * (x)|_{\frac{y}{5}}^{\frac{y^{1/2}}{3}}\,dx$

$f_Y(y) = \frac{2187}{625} * y * (\frac{y^{1/2}}{3} - \frac{y}{5})$

$f_Y(y) = \frac{2187}{625} * (\frac{y^{3/2}}{3} - \frac{y^2}{5})$

Entonces, la función Densidad marginal de $f_Y(y)$ queda definida como:

$
f_Y(y) = \begin{cases}
\frac{2187}{625} * (\frac{y^{3/2}}{3} - \frac{y^2}{5}) & \text{si } 0 \le y \le \frac{25}{9} \\
0 & \text{en otro caso}
\end{cases}
$

### Paso 4 - Definimos la función Densidad condicional $f_{X|Y}(x|y)$ de $X$ dado $Y$

Por definición, la función Densidad condicional $f_{X|Y}(x|y)$ de $X$ dado $Y$ es:

$f_{X|Y}(x|y) = \frac{f_{X,Y}(x,y)}{f_Y(y)}$

Siendo $f_{X,Y}(x,y)$ la Función Densidad Conjunta (ver Paso 2 de este ejercicio):

$
f_{X,Y} = \begin{cases}
\frac{2187}{625}y & \text{si } 9x^{2} \le y \le 5x &;& 0 \le x \le \frac{5}{9} \\
0 & \text{en otro caso}
\end{cases}
$

Y siendo $f_Y(y)$ la Función Densidad Marginal (ver Paso 3 de este ejercicio):

$
f_Y(y) = \begin{cases}
\frac{2187}{625} * (\frac{y^{3/2}}{3} - \frac{y^2}{5}) & \text{si } 0 \le y \le \frac{25}{9} \\
0 & \text{en otro caso}
\end{cases}
$

Calculamos la función Densidad condicional $f_{X|Y}(x|y)$ de $X$ dado $Y$:

$f_{X|Y}(x|y) = \frac{f_{X,Y}(x,y)}{f_Y(y)}$

$f_{X|Y}(x|y) = \frac{2187}{625}y * \frac{1}{\frac{2187}{625} * (\frac{y^{3/2}}{3} - \frac{y^2}{5})}$

$f_{X|Y}(x|y) = y * \frac{1}{\frac{y^{3/2}}{3} - \frac{y^2}{5}}$

$f_{X|Y}(x|y) = y * \frac{1}{\frac{5y^{3/2}-3y^{2}}{15}}$

$f_{X|Y}(x|y) = \frac{15y}{5y^{3/2}-3y^{2}}$

Finalmente, tenemos:

$
f_{X|Y}(x|y) = \begin{cases}
\frac{15y}{5y^{3/2}-3y^{2}} & \text{si } \frac{y}{5} \le x \le \frac{y^\frac{1}{2}}{3} &;& \text{para } 0 \lt y \lt \frac{25}{9} \text{ fijo}\\
0 & \text{en otro caso}
\end{cases}
$

### Paso 2 - Gráfico de Función Empírica de Distribución y Función de Densidad de Ventas por año

En el siguiente paso graficamos la funciones `Función Empírica de Distribución` (a la izquierda) y `Función de Densidad de Probabilidad` (a la derecha), de las Ventas ***por año***.
